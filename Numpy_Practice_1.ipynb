```python
# ---
# ## Section 1 — Practice Questions

# Solve the problems in the code cell below.

# **Q1.** Create a Python list `[5, 10, 15, 20, 25]`. 

l = [5,10,15,20,25]

# Print the **3rd element** and the **last element**.

print(l[2]) 
print(l[-1]) 

# **Q2.** Write a loop that prints the **square** of each number in `[1, 2, 3, 4, 5]`.  

sq = []
for i in l:
    sq.append(i**2)

print(sq)

# Then rewrite the same thing using a **list comprehension** in one line.

sq = [i**2 for i in l]

# **Q3.** Convert the list `[3, 6, 9, 12, 15]` to a NumPy array.  

arr = np.array([3,6,9,12,15])

# Multiply every element by `4` using a **single operation** (no loop).

arr = arr * 4
arr
# ---
# > **Remember:** In NumPy, one line replaces many lines of looping code!
```

```python
# ---
# ## Section 3 — Practice Questions
# **Q1.** Create a 2D array of shape `(4, 5)` using `np.arange(20).reshape(4, 5)`.  

arr_2d = np.arange(20).reshape(4,5)
print(arr_2d)

# Print all six attributes: `.ndim`, `.shape`, `.size`, `.dtype`, `.itemsize`, `.nbytes`.
print("ndim:", arr_2d.ndim)
print("shape:", arr_2d.shape)
print("size:", arr_2d.size)
print("dtype:", arr_2d.dtype)
print("itemsize:", arr_2d.itemsize)
print("nbytes:", arr_2d.nbytes)


# **Q2.** Create two arrays with the **same values** but different dtypes: `float64` and `float32`.  
arr_float64 = np.array([1.0, 2.0, 3.0], dtype=np.float64)
arr_float32 = np.array([1.0, 2.0, 3.0], dtype=np.float32)

# Compare their `.itemsize` and `.nbytes`. How much memory do you save with `float32`?
print("float64 itemsize:", arr_float64.itemsize)
print("float64 nbytes:", arr_float64.nbytes)
print("float32 itemsize:", arr_float32.itemsize)
print("float32 nbytes:", arr_float32.nbytes)
memory_saved = arr_float64.nbytes - arr_float32.nbytes
print(f"Memory saved by using float32: {memory_saved} bytes")


# **Q3.** A dataset has **10,000 samples** and **256 features**, stored as `float32`.  
# Without creating the array, **calculate** the memory it would use in MB.  
# Then verify by creating the array with `np.zeros((10000, 256), dtype=np.float32)`.
num_samples = 10000
num_features = 256
dtype_size = np.dtype(np.float32).itemsize
memory_bytes = num_samples * num_features * dtype_size
memory_mb = memory_bytes / (1024 ** 2)
print(f"Estimated memory usage: {memory_mb} MB")

arr_dataset = np.zeros((10000, 256), dtype=np.float32)
actual_memory_mb = arr_dataset.nbytes / (1024 ** 2)
print(f"Actual memory usage: {actual_memory_mb} MB")


# > **Real-world note:** In deep learning, models store millions of weights as `float32`.  
# > Knowing `.nbytes` helps you estimate whether a model fits in GPU memory!



# ## Core Attributes at a Glance
# | Attribute | Returns | Meaning |
# |-----------|---------|---------|
# | `.ndim` | integer | Number of dimensions (axes) |
# | `.shape` | tuple | Size along each dimension |
# | `.size` | integer | Total number of elements |
# | `.dtype` | dtype object | Data type of elements |
# | `.itemsize` | integer | Bytes used by **one** element |
# | `.nbytes` | integer | Total bytes used by the array |
```
